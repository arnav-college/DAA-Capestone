## Problem Statement

A courier company receives multiple parcels that must be delivered to different customer locations. Each parcel has a destination, weight, delivery deadline, priority level, and delivery value.

Because the delivery vehicle has limited carrying capacity, the system must choose which parcels should be loaded for one trip. After parcel selection, the chosen destinations can be used for route planning.

## Objective

Design a courier-planning prototype that can:

- Store parcel details.
- Prioritize parcels by priority and deadline.
- Select the best parcels within vehicle capacity.
- Represent delivery locations as a graph for shortest-path and route planning.

## Data Design

| Attribute | Meaning |
|---|---|
| parcel_id | Unique parcel identifier |
| destination | Delivery location code |
| weight | Parcel weight in kg |
| value | Business value or delivery reward |
| deadline | Required delivery day; smaller means more urgent |
| priority | Priority category: High, Medium, or Low |
| benefit | Combined score used for optimization |

```python
from dataclasses import dataclass
from typing import List, Tuple


@dataclass(frozen=True)
class Parcel:
    parcel_id: str
    destination: str
    weight: int
    value: int
    deadline: int
    priority: str


PRIORITY_SCORE = {
    "High": 3,
    "Medium": 2,
    "Low": 1,
}


def priority_score(parcel: Parcel) -> int:
    return PRIORITY_SCORE[parcel.priority]


def parcel_benefit(parcel: Parcel) -> int:
    urgency_bonus = max(0, 6 - parcel.deadline) * 5
    priority_bonus = priority_score(parcel) * 20
    return parcel.value + priority_bonus + urgency_bonus


parcels = [
    Parcel("P01", "A", 4, 60, 2, "High"),
    Parcel("P02", "B", 2, 35, 1, "High"),
    Parcel("P03", "C", 6, 80, 4, "Medium"),
    Parcel("P04", "D", 3, 45, 3, "High"),
    Parcel("P05", "E", 5, 70, 2, "Medium"),
    Parcel("P06", "F", 1, 20, 1, "Low"),
    Parcel("P07", "G", 7, 100, 5, "High"),
    Parcel("P08", "H", 2, 25, 3, "Low"),
    Parcel("P09", "I", 2, 30, 3, "High"),
]

VEHICLE_CAPACITY = 12
```

## Graph Design

The delivery area is represented as a weighted graph:

- Each delivery point is a graph node.
- Each road is a graph edge.
- Each edge weight stores distance or travel time.
- Dijkstra's algorithm can find shortest paths.
- A nearest-neighbor heuristic can create a practical delivery order.

## Algorithm Plan

| Component | Algorithm | Purpose |
|---|---|---|
| Parcel prioritization | Stable sorting | Order parcels by priority and urgency |
| Box filling | 0/1 knapsack dynamic programming | Select the best parcel set within vehicle capacity |
| Shortest paths | Dijkstra's algorithm | Find shortest road paths |
| Route planning | Nearest-neighbor heuristic | Build an approximate delivery route |
